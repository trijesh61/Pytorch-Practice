## Batch Gradient descent

Let's train 

```python
w1=torch.rand(1,requires_grad=True)
w2=torch.rand(1,requires_grad=True)
w3=torch.rand(1,requires_grad=True)
bias=torch.rand(1,requires_grad=True)

learning_rate=0.006
epochs=5000
for epoch in range(epochs):
    prediccted_bonus=w1*performance+w2*years_of_experience+w3*projects_completed+bias
    loss=((prediccted_bonus-bonus)**2).mean()
    loss.backward()
    with torch.no_grad():
        w1-=learning_rate*w1.grad
        w2-=learning_rate*w2.grad
        w3-=learning_rate*w3.grad
        bias-=learning_rate*bias.grad
    w1.grad.zero_()
    w2.grad.zero_()
    w3.grad.zero_()
    bias.grad.zero_()
    if (epoch+1)%100==0:
        print(f"Epoch  :  {epoch}  ,   Loss : {loss.item():0.2f}") 
```

```python
# Step 4: Print the learned weights and bias
print(f"\nLearned weights and bias:")
print(f"w1: {w1.item():.2f}, w2: {w2.item():.2f}, w3: {w3.item():.2f}, bias: {bias.item():.2f}")
```

## Mini - Batch Gradient Descent

```python
w1=torch.rand(1,requires_grad=True)
w2=torch.rand(1,requires_grad=True)
w3=torch.rand(1,requires_grad=True)
bias=torch.rand(1,requires_grad=True)

n_samples=len(performance)
batch_size=16
learning_rate=0.001
epochs=1000
for epoch in range(epochs):
    for i in range(0,n_samples,batch_size):
        batch_performance=performance[i:i+batch_size]
        batch_years_of_experience=years_of_experience[i:i+batch_size]
        batch_projects_completed=projects_completed[i:i+batch_size]
        batch_bonus=bonus[i:i+batch_size]

        predicted_batch_bonus=w1*batch_performance+w2*batch_years_of_experience+w3*batch_projects_completed+bias
        loss=((predicted_batch_bonus-batch_bonus)**2).mean()
        loss.backward()
        with torch.no_grad():
            w1-=learning_rate*w1.grad
            w2-=learning_rate*w2.grad
            w3-=learning_rate*w3.grad
            bias-=learning_rate*bias.grad
        
    w1.grad.zero_()
    w2.grad.zero_()
    w3.grad.zero_()
    bias.grad.zero_()

    if (epoch+1)%100==0:
        print(f"Epoch  :  {epoch+1}  ,   Loss : {loss.item():0.2f}") 
```

## Stochastic Gradient Descent

```python
w1 = torch.rand(1, requires_grad=True)
w2 = torch.rand(1, requires_grad=True)
w3 = torch.rand(1, requires_grad=True)
bias = torch.rand(1, requires_grad=True)

n_samples = len(performance)
learning_rate = 0.001
epochs = 500

for epoch in range(epochs):

    # Shuffle indices (important for SGD)
    indices = torch.randperm(n_samples)

    for i in indices:
        x1 = performance[i]
        x2 = years_of_experience[i]
        x3 = projects_completed[i]
        y = bonus[i]

        # Forward pass
        predicted_bonus = w1 * x1 + w2 * x2 + w3 * x3 + bias
        loss = (predicted_bonus - y) ** 2

        # Backward pass
        loss.backward()

        # Update weights immediately (SGD)
        with torch.no_grad():
            w1 -= learning_rate * w1.grad
            w2 -= learning_rate * w2.grad
            w3 -= learning_rate * w3.grad
            bias -= learning_rate * bias.grad

        # Zero gradients after every sample
        w1.grad.zero_()
        w2.grad.zero_()
        w3.grad.zero_()
        bias.grad.zero_()

    if epoch % 100 == 0:
        print(f"Epoch: {epoch}, Loss: {loss.item():.4f}")
```

```python
Age=torch.tensor(df["Age"].values,dtype=torch.float32)
Dependants=torch.tensor(df["Dependants"].values,dtype=torch.float32)
Income_Lakhs=torch.tensor(df["Income_Lakhs"].values,dtype=torch.float32)
Genetical_Risk=torch.tensor(df["Genetical_Risk"].values,dtype=torch.float32)
Cost=torch.tensor(df["Cost"].values,dtype=torch.float32)
```

```python
from torch import no_grad


w1=torch.rand(1,requires_grad=True)
w2=torch.rand(1,requires_grad=True)
w3=torch.rand(1,requires_grad=True)
w4=torch.rand(1,requires_grad=True)
bias=torch.rand(1,requires_grad=True)
epochs=1000
learning_rate=0.0001
for epoch in range(epochs):
    predicted_cost=w1*Age+w2*Dependants+w3*Income_Lakhs+w4*Genetical_Risk+bias
    loss=((predicted_cost-Cost)**2).mean()

    loss.backward()
    with torch.no_grad():
        w1-=learning_rate*w1.grad
        w2-=learning_rate*w2.grad
        w3-=learning_rate*w3.grad
        w4-=learning_rate*w4.grad
        bias-=learning_rate*bias.grad
    w1.grad.zero_()
    if (epoch+1)%100 == 0:
        print(f"Epochs:{epoch}, Loss:  {loss.item():0.2f}")
```
